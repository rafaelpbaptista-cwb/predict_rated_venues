# Carregamento e Preparação do Dataset de Testes

Verifica-se se o pipeline de pré-processamento estruturado na etapa de treinamento aplica as transformações necessárias sobre o conjunto de teste (Kaggle), mantendo a mesma estrutura esperada pelos classificadores.

```python
from carga_tratamento_dados import carregar_dados

df_testes = carregar_dados(carregar_train=False, div_dataset_validacao=False)

df_testes.head()
```

## Predição com o Modelo Selecionado

```python
import pickle
from datetime import datetime

with open("data/models/random_forest_model.pkl", "rb") as file:
    model_final = pickle.load(file)

predict = model_final.predict(df_testes.iloc[:, :-1])

df_resultado = df_testes.reset_index()["business_id"].to_frame()
df_resultado["destaque"] = predict

df_resultado.to_csv(f"resultado_{datetime.now().strftime('%Y%m%d_%H%M')}.csv", index=False)
```
